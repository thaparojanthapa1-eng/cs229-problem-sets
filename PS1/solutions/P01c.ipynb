## Derivation of Gaussian Discriminant Analysis (GDA) Posterior Probability

$$P(y=1 \mid x; \phi, \mu_0, \mu_1, \Sigma)$$

$$= \frac{P(x \mid y=1) P(y=1)}{P(x \mid y=1) P(y=1) + P(x \mid y=0) P(y=0)}$$

$$= \frac{\left[ \frac{1}{(2\pi)^{n/2} |\Sigma|^{1/2}} \exp\left( -\frac{1}{2} (x - \mu_1)^T \Sigma^{-1} (x - \mu_1) \right) \right] \phi}{\left[ \frac{1}{(2\pi)^{n/2} |\Sigma|^{1/2}} \exp\left( -\frac{1}{2} (x - \mu_1)^T \Sigma^{-1} (x - \mu_1) \right) \right] \phi + \left[ \frac{1}{(2\pi)^{n/2} |\Sigma|^{1/2}} \exp\left( -\frac{1}{2} (x - \mu_0)^T \Sigma^{-1} (x - \mu_0) \right) \right] (1 - \phi)}$$

Dividing the numerator and denominator by $\frac{1}{(2\pi)^{n/2} |\Sigma|^{1/2}}$:

$$= \frac{\exp\left( -\frac{1}{2} (x - \mu_1)^T \Sigma^{-1} (x - \mu_1) \right) \phi}{\exp\left( -\frac{1}{2} (x - \mu_1)^T \Sigma^{-1} (x - \mu_1) \right) \phi + \exp\left( -\frac{1}{2} (x - \mu_0)^T \Sigma^{-1} (x - \mu_0) \right) (1 - \phi)}$$

Dividing the numerator and denominator by $\exp\left( -\frac{1}{2} (x - \mu_1)^T \Sigma^{-1} (x - \mu_1) \right) \phi$:

$$= \frac{1}{1 + \exp\left[ \frac{1}{2} (x - \mu_1)^T \Sigma^{-1} (x - \mu_1) - \frac{1}{2} (x - \mu_0)^T \Sigma^{-1} (x - \mu_0) \right] \left( \frac{1 - \phi}{\phi} \right)}$$

Expanding and simplifying inside the exponential function:

$$= \frac{1}{1 + \exp\left\{ - \left[ \Sigma^{-1} (\mu_1 - \mu_0) \right]^T x + \frac{1}{2} (\mu_0 + \mu_1)^T \Sigma^{-1} (\mu_0 - \mu_1) - \ln\left( \frac{1 - \phi}{\phi} \right) \right\}}$$

---

### Comparison to the Sigmoid Function

Comparing this to the logistic function form:

$$\frac{1}{1 + e^{-z}}$$

where:

$$z = \theta^T x + \theta_0$$

We get the parameters:

$$\theta^T = \left( \Sigma^{-1} (\mu_1 - \mu_0) \right)^T$$

$$\theta_0 = \frac{1}{2} (\mu_0 + \mu_1)^T \Sigma^{-1} (\mu_0 - \mu_1) - \ln\left( \frac{1 - \phi}{\phi} \right)$$
