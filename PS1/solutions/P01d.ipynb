# Gaussian Discriminant Analysis (GDA) Parameter Estimation via MLE

$$\phi = \frac{1}{m} \sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\}$$

$$\mu_0 = \frac{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\} x^{(i)}}{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\}}$$

$$\mu_1 = \frac{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\} x^{(i)}}{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\}}$$

$$\Sigma = \frac{1}{m} \sum_{i=1}^{m} (x^{(i)} - \mu_{y^{(i)}})(x^{(i)} - \mu_{y^{(i)}})^T$$

---

### Probabilistic Models

$$P(y^{(i)}; \phi) = \phi^{y^{(i)}} (1 - \phi)^{1 - y^{(i)}}$$

$$P(x^{(i)} \mid y^{(i)} = 0; \mu_0, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x^{(i)} - \mu_0)^2}{2\sigma^2} \right)$$

$$P(x^{(i)} \mid y^{(i)} = 1; \mu_1, \sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x^{(i)} - \mu_1)^2}{2\sigma^2} \right)$$

---

### Log-Likelihood Formulation

$$l(\phi, \mu_0, \mu_1, \Sigma) = \log \prod_{i=1}^{m} P(x^{(i)} \mid y^{(i)}; \mu_0, \mu_1, \Sigma) P(y^{(i)}; \phi)$$

$$= \sum_{i=1}^{m} \log P(y^{(i)}; \phi) + \sum_{i=1}^{m} \log P(x^{(i)} \mid y^{(i)}; \mu_0, \mu_1, \Sigma)$$

$$= \sum_{i=1}^{m} \left[ y^{(i)} \log \phi + (1 - y^{(i)}) \log(1 - \phi) \right] + \sum_{i=1}^{m} \left[ -\frac{1}{2}\log(2\pi) - \frac{1}{2}\log(\sigma^2) - \mathbb{I}\{y^{(i)} = 0\}\frac{(x^{(i)} - \mu_0)^2}{2\sigma^2} - \mathbb{I}\{y^{(i)} = 1\}\frac{(x^{(i)} - \mu_1)^2}{2\sigma^2} \right]$$

---

### Parameter Derivations

#### 1. Derivation for $\phi$

$$\frac{\partial l}{\partial \phi} = \sum_{i=1}^{m} \left( \frac{y^{(i)}}{\phi} - \frac{1 - y^{(i)}}{1 - \phi} \right) = 0$$

$$\frac{1}{\phi} \sum_{i=1}^{m} y^{(i)} = \frac{1}{1 - \phi} \sum_{i=1}^{m} (1 - y^{(i)})$$

$$(1 - \phi) \sum_{i=1}^{m} y^{(i)} = \phi \sum_{i=1}^{m} (1 - y^{(i)})$$

$$\sum_{i=1}^{m} y^{(i)} - \phi \sum_{i=1}^{m} y^{(i)} = \phi m - \phi \sum_{i=1}^{m} y^{(i)}$$

$$\phi = \frac{1}{m} \sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\}$$

---

#### 2. Derivation for $\mu_0$ and $\mu_1$

$$\frac{\partial l}{\partial \mu_0} = \frac{\partial}{\partial \mu_0} \left[ -\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\} \frac{(x^{(i)} - \mu_0)^2}{2\sigma^2} \right] = 0$$

$$\frac{1}{\sigma^2} \sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\}(x^{(i)} - \mu_0) = 0$$

$$\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\} x^{(i)} - \mu_0 \sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\} = 0$$

$$\mu_0 = \frac{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\} x^{(i)}}{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 0\}}$$

Similarly, setting $\frac{\partial l}{\partial \mu_1} = 0$:

$$\mu_1 = \frac{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\} x^{(i)}}{\sum_{i=1}^{m} \mathbb{I}\{y^{(i)} = 1\}}$$

---

#### 3. Derivation for Variance $\sigma^2$ (and Covariance $\Sigma$)

$$l(\sigma^2) = -\frac{m}{2} \log(\sigma^2) - \frac{1}{2\sigma^2} \sum_{i=1}^{m} (x^{(i)} - \mu_{y^{(i)}})^2$$

$$\frac{\partial l}{\partial \sigma^2} = -\frac{m}{2\sigma^2} + \frac{1}{2(\sigma^2)^2} \sum_{i=1}^{m} (x^{(i)} - \mu_{y^{(i)}})^2 = 0$$

$$-m + \frac{1}{\sigma^2} \sum_{i=1}^{m} (x^{(i)} - \mu_{y^{(i)}})^2 = 0$$

$$\sigma^2 = \frac{1}{m} \sum_{i=1}^{m} (x^{(i)} - \mu_{y^{(i)}})^2$$

Generalizing to the multivariate covariance matrix:

$$\Sigma = \frac{1}{m} \sum_{i=1}^{m} (x^{(i)} - \mu_{y^{(i)}})(x^{(i)} - \mu_{y^{(i)}})^T$$
