# Logistic Regression: Gradient and Hessian Derivation

---

## 1. Cost Function

$$J(\theta) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \log\left(h_\theta(x^{(i)})\right) + \left(1 - y^{(i)}\right) \log\left(1 - h_\theta(x^{(i)})\right) \right]$$

---

## 2. Derivative of the Hypothesis Function

$$\frac{\partial h_\theta(x^{(i)})}{\partial \theta_j} = g'(\theta^T x^{(i)}) x_j^{(i)} = g(\theta^T x^{(i)}) \left(1 - g(\theta^T x^{(i)})\right) x_j^{(i)}$$

---

## 3. Derivation of the Gradient $\frac{\partial J(\theta)}{\partial \theta_j}$

$$\frac{\partial J(\theta)}{\partial \theta_j} = -\frac{1}{m} \sum_{i=1}^{m} \left[ \frac{y^{(i)}}{h_\theta(x^{(i)})} \cdot h_\theta(x^{(i)})\left(1 - h_\theta(x^{(i)})\right) x_j^{(i)} - \frac{1 - y^{(i)}}{1 - h_\theta(x^{(i)})} \cdot h_\theta(x^{(i)})\left(1 - h_\theta(x^{(i)})\right) x_j^{(i)} \right]$$

$$= -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} \left(1 - h_\theta(x^{(i)})\right) x_j^{(i)} - (1 - y^{(i)}) h_\theta(x^{(i)}) x_j^{(i)} \right]$$

$$= -\frac{1}{m} \sum_{i=1}^{m} \left[ y^{(i)} - y^{(i)} h_\theta(x^{(i)}) - h_\theta(x^{(i)}) + y^{(i)} h_\theta(x^{(i)}) \right] x_j^{(i)}$$

$$= -\frac{1}{m} \sum_{i=1}^{m} \left( y^{(i)} - h_\theta(x^{(i)}) \right) x_j^{(i)}$$

$$= \frac{1}{m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right) x_j^{(i)}$$

---

## 4. Hessian Matrix Elements $H_{jk}$

$$H_{jk} = \frac{\partial}{\partial \theta_k} \left[ \frac{1}{m} \sum_{i=1}^{m} \left( h_\theta(x^{(i)}) - y^{(i)} \right) x_j^{(i)} \right]$$

$$= \frac{1}{m} \sum_{i=1}^{m} \frac{\partial h_\theta(x^{(i)})}{\partial \theta_k} x_j^{(i)}$$

$$= \frac{1}{m} \sum_{i=1}^{m} h_\theta(x^{(i)}) \left(1 - h_\theta(x^{(i)})\right) x_j^{(i)} x_k^{(i)}$$

---

## 5. Convexity Proof ($z^T H z \ge 0$)

$$z^T H z = z^T \left[ \frac{1}{m} \sum_{i=1}^{m} h_\theta(x^{(i)})\left(1 - h_\theta(x^{(i)})\right) x_j^{(i)} x_k^{(i)} \right] z$$

$$= \frac{1}{m} \sum_{i=1}^{m} \sum_{j=1}^{n} \sum_{k=1}^{n} g(\theta^T x^{(i)}) \left(1 - g(\theta^T x^{(i)})\right) x_j^{(i)} x_k^{(i)} z_j z_k$$

$$= \frac{1}{m} \sum_{i=1}^{m} g(\theta^T x^{(i)}) \left(1 - g(\theta^T x^{(i)})\right) \left( {x^{(i)}}^T z \right)^2$$

$$z^T H z \ge 0$$
