### **Proof: Variance of Sufficient Statistics in Exponential Families**

Starting from the expected score function equation derived previously:

$$\int p(y; \eta) \left( T(y) - \nabla_\eta a(\eta) \right) dy = 0$$

Taking the gradient with respect to $\eta$ on both sides:

$$\nabla_\eta \int p(y; \eta) \left( T(y) - \nabla_\eta a(\eta) \right) dy = 0$$

Swapping integration and differentiation:

$$\int \nabla_\eta \left[ p(y; \eta) \left( T(y) - \nabla_\eta a(\eta) \right) \right] dy = 0$$

---

#### **Applying the Product Rule**

Using the matrix product rule $\nabla_\eta [u v^\top] = (\nabla_\eta u) v^\top + u (\nabla_\eta v^\top)$:

$$
\begin{aligned}
\nabla_\eta \left[ p(y; \eta) \left( T(y) - \nabla_\eta a(\eta) \right)^\top \right] &= \left( \nabla_\eta p(y; \eta) \right) \left( T(y) - \nabla_\eta a(\eta) \right)^\top \\
&\quad + p(y; \eta) \nabla_\eta \left( T(y) - \nabla_\eta a(\eta) \right)^\top
\end{aligned}
$$

Substitute the known components:
1. $\nabla_\eta p(y; \eta) = p(y; \eta) \left( T(y) - \nabla_\eta a(\eta) \right)$
2. $\nabla_\eta \left( T(y) - \nabla_\eta a(\eta) \right)^\top = -\nabla_\eta^2 a(\eta)$

---

#### **Plugging Back into Integral**

Substituting these expressions back into the integral gives:

$$\int p(y; \eta) \left( T(y) - \nabla_\eta a(\eta) \right) \left( T(y) - \nabla_\eta a(\eta) \right)^\top dy - \int p(y; \eta) \nabla_\eta^2 a(\eta) \, dy = 0$$

Since $\nabla_\eta a(\eta) = \mathbb{E}[T(y)]$, the first integral becomes the definition of variance:

$$\mathbb{E} \left[ \left( T(y) - \mathbb{E}[T(y)] \right) \left( T(y) - \mathbb{E}[T(y)] \right)^\top \right] = \text{Var}(T(y))$$

---

#### **Final Simplification**

Substituting this back:

$$\text{Var}(T(y)) - \nabla_\eta^2 a(\eta) \cdot \int p(y; \eta) \, dy = 0$$

Since $\int p(y; \eta) \, dy = 1$:

$$\text{Var}(T(y)) = \nabla_\eta^2 a(\eta)$$

$$\therefore \text{Var}(Y \mid X; \theta) = \nabla_\eta^2 a(\eta)$$
