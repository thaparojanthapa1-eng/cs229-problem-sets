### **Proof: Expectation of Sufficiency Statistics in Exponential Families**

#### **4. a.**

Starting with the total probability axiom:

$$\int p(y; \eta) \, dy = 1$$

Taking the gradient with respect to $\eta$ on both sides:

$$\nabla_\eta \int p(y; \eta) \, dy = \nabla_\eta (1) = 0$$

---

#### **Gradient of the Probability Density Function**

Given the exponential family distribution format $p(y; \eta) = b(y) \exp\left(\eta^\top T(y) - a(\eta)\right)$:

$$
\begin{aligned}
\nabla_\eta p(y; \eta) &= \nabla_\eta \left[ b(y) \exp\left(\eta^\top T(y) - a(\eta)\right) \right] \\
&= b(y) \exp\left(\eta^\top T(y) - a(\eta)\right) \cdot \nabla_\eta \left( \eta^\top T(y) - a(\eta) \right) \\
&= p(y; \eta) \cdot \left( T(y) - \nabla_\eta a(\eta) \right)
\end{aligned}
$$

---

#### **Deriving the Expectation**

Substituting the gradient expression back into the integral:

$$\therefore \int p(y; \eta) \cdot \left( T(y) - \nabla_\eta a(\eta) \right) dy = 0$$

Expanding and splitting the integral:

$$\int T(y) \, p(y; \eta) \, dy - \int (\nabla_\eta a(\eta)) \, p(y; \eta) \, dy = 0$$

Factoring out terms independent of $y$:

$$\int T(y) \, p(y; \eta) \, dy - \nabla_\eta a(\eta) \int p(y; \eta) \, dy = 0$$

Since $\int p(y; \eta) \, dy = 1$, we simplify to the expectation definition:

$$E[T(y)] - \nabla_\eta a(\eta) = 0$$

$$E[T(y)] = \nabla_\eta a(\eta)$$

$$\therefore E[Y \mid X; \theta] = \nabla_\eta a(\eta)$$
