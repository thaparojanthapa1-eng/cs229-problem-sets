### **Proof: Convexity of NLL Loss and Positive Semi-Definiteness of the Hessian**

#### **c)** Probability Density & Log-Likelihood

For an Exponential Family distribution:

$$p(y \mid x; \theta) = b(y) \exp\left( \eta^\top T(y) - a(\eta) \right)$$

where $\eta = x_i^\top \theta$.

The log-likelihood function $l(\theta)$ is:

$$
\begin{aligned}
l(\theta) &= \sum_{i=1}^{N} \log p(y_i \mid x_i; \theta) \\
&= \sum_{i=1}^{N} \left[ \log b(y_i) + (x_i^\top \theta) T(y_i) - a(x_i^\top \theta) \right]
\end{aligned}
$$

---

#### **Negative Log-Likelihood (NLL) Loss**

The NLL loss function $\mathcal{L}(\theta)$ is defined as $-l(\theta)$:

$$\mathcal{L}(\theta) = \sum_{i=1}^{N} \left[ a(x_i^\top \theta) - (x_i^\top \theta) T(y_i) - \log b(y_i) \right]$$

---

#### **Gradient of the NLL Loss**

Note that $\nabla_\theta (x_i^\top \theta) = x_i$. Applying the chain rule to $a(x_i^\top \theta)$:

$$
\begin{aligned}
\nabla_\theta a(x_i^\top \theta) &= \left( \frac{\partial \eta_i}{\partial \theta} \right)^\top \nabla_{\eta_i} a(\eta_i) \\
&= x_i \, \nabla_{\eta_i} a(\eta_i) \\
&= x_i \, \mathbb{E}[T(y_i) \mid x_i]
\end{aligned}
$$

Taking the gradient of the loss $\mathcal{L}(\theta)$ with respect to $\theta$:

$$
\begin{aligned}
\nabla_\theta \mathcal{L}(\theta) &= \sum_{i=1}^{N} x_i \left( \nabla_{\eta_i} a(\eta_i) - T(y_i) \right) \\
&= \sum_{i=1}^{N} x_i \left( \mathbb{E}[T(y_i) \mid x_i] - T(y_i) \right)
\end{aligned}
$$

---

#### **Hessian of the NLL Loss Function**

The Hessian $H_\theta \mathcal{L}(\theta)$ is the gradient of the transpose of the loss gradient:

$$H_\theta \mathcal{L}(\theta) = \nabla_\theta \left[ \nabla_\theta \mathcal{L}(\theta) \right]^\top = \sum_{i=1}^{N} x_i \left( \nabla_\theta \left[ \nabla_{\eta_i} a(\eta_i) \right] \right)^\top$$

Applying the chain rule to $\nabla_{\eta_i} a(\eta_i)$:

$$\nabla_\theta \left[ \nabla_{\eta_i} a(\eta_i) \right]^\top = \left( \nabla_{\eta_i}^2 a(\eta_i) \right) \left( \frac{\partial \eta_i}{\partial \theta} \right)^\top = \nabla_{\eta_i}^2 a(\eta_i) \, x_i^\top$$

Substituting this back gives the final Hessian form:

$$H_\theta \mathcal{L}(\theta) = \sum_{i=1}^{N} x_i \left( \nabla_{\eta_i}^2 a(\eta_i) \right) x_i^\top$$

---

#### **Proof of Positive Semi-Definiteness**

From the variance property of Exponential Families:

$$\nabla_\eta^2 a(\eta) = \text{Var}(T(y) \mid x)$$

Since variance is always non-negative:

$$\nabla_\eta^2 a(\eta) \ge 0$$

To test for positive semi-definiteness, evaluate $v^\top H_\theta \mathcal{L}(\theta) v$ for any arbitrary non-zero vector $v \in \mathbb{R}^d$:

$$
\begin{aligned}
v^\top H_\theta \mathcal{L}(\theta) v &= v^\top \left( \sum_{i=1}^{N} x_i \left( \nabla_{\eta_i}^2 a(\eta_i) \right) x_i^\top \right) v \\
&= \sum_{i=1}^{N} (v^\top x_i) \left( \nabla_{\eta_i}^2 a(\eta_i) \right) (x_i^\top v) \\
&= \sum_{i=1}^{N} (x_i^\top v)^\top \left( \nabla_{\eta_i}^2 a(\eta_i) \right) (x_i^\top v)
\end{aligned}
$$

Let $z_i = x_i^\top v$ (a scalar), so $z_i^2 = (x_i^\top v)^2 \ge 0$. Then:

$$v^\top H_\theta \mathcal{L}(\theta) v = \sum_{i=1}^{N} z_i^2 \, \text{Var}(T(y) \mid x)$$

Since $z_i^2 \ge 0$ and $\text{Var}(T(y) \mid x) \ge 0$, every term in the sum is non-negative:

$$v^\top H_\theta \mathcal{L}(\theta) v \ge 0 \quad \forall v \in \mathbb{R}^d$$

$$\therefore H_\theta \mathcal{L}(\theta) \text{ is Positive Semi-Definite (PSD), proving } \mathcal{L}(\theta) \text{ is convex.}$$
